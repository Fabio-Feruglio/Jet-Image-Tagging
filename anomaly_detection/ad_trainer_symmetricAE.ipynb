## 2a. Symmetric autoencoder (no noise) training

```python
# Train the Standard Symmetric Autoencoder (No Noise)
!python /content/Jet-Image-Tagging/anomaly_detection/src/train_symmetricAE.py \
    --data_path "/content/jet_images_128.h5" \
    --save_dir "/content/drive/MyDrive/JetTagging/anomaly_detection/checkpoints_ae_standard" \
    --epochs 40 \
    --latent_space_dim 16 \
    --batch_size 128 \
    --lr 0.0005 \
    --bg_classes 0 1 \
    --img_size 128 \
    --weight_decay 0.0001 \
    --max_samples 30000 \
    --patience 10 \
    --noise_prob 0.0
```

```python
# Evaluation - Standard Autoencoder
!python /content/Jet-Image-Tagging/anomaly_detection/src/evaluation_symmetricAE.py \
    --model_path "/content/drive/MyDrive/JetTagging/anomaly_detection/checkpoints/symmetricAE/autoencoder_best.pth" \
    --max_samples 30000 \
    --latent_space_dim 16 \
    --batch_size 128 \
    --img_size 128 \
    --bg_classes 0 1 \
    --data_path "/content/jet_images_128.h5" \
    --save_dir "/content/drive/MyDrive/JetTagging/anomaly_detection/results/symmetricAE"
```
